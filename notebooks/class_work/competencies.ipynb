```python

import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("competencies.csv")
effort=df.groupby("Student")["Grade"].mean()


plt.figure(figsize=(8,8))

effort.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%') 

plt.title("Who performs best?")
plt.ylabel("Grade")
plt.show
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Does more studying = better grades?")
plt.xlabel("Hours studied")
plt.ylabel("Grades")
plt.show()
```

```python
Hours_min=df["Hours_Studied"].min()
Hours_q1=df["Hours_Studied"].quantile(0.25)
print("Minimum hours: ", Hours_min, "Q1: ", Hours_q1)

plt.figure(figsize=(8,6))
df[["Hours_Studied","Grade"]].boxplot()

plt.title("how consistent are study patterns and performance?")
plt.xlabel("")
plt.ylabel("Value")
plt.show()
```
