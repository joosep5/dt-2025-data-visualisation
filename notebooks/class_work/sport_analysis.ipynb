```python
import pandas as pd

df= pd.read_csv("sports.csv")
df.head() # this shows first rows
```

```python
df.describe() # generates some statistics
```

Box plot for each athlete to find the number of years of practice

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years of practice", legend=False)
plt.show
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years of practice", legend=False)
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.ylabel("Age")
plt.title("Age per athlete")
plt.show
```

```python
plt.scatter(df["Age"], df["Years of practice"], s=df["Years of practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs years of practice")
plt.show
```

```python
avg_age=df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Avrage age per sports")
plt.xlabel("Sports")
plt.ylabel("Age")
plt.show()
```
