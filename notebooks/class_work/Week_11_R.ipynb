```python
import urllib.request

# --- Read text from URL ---
text = urllib.request.urlopen(
    "https://drive.google.com/uc?export=download&id=1f6RErkEKe_WqcS5LmYEm8Zi81lR1cmk0"
).read().decode("utf-8")

# --- Download an image from URL (optional, from your code) ---
img = urllib.request.urlretrieve(
    "https://drive.google.com/uc?export=download&id=1o19a6yrRM6ikHr3FytSICT8nfdDA1qhK",
    "image1.png"
)

# --- Text operations ---
# 1. Split text into words
words = text.split()

# 2. Count total words
word_count = len(words)

# 3. Select the first 10 words
first_10_words = words[:10]

# 4. Find all words that start with a capital letter
capital_words = [w for w in words if w.istitle()]

# 5. Print results
print("Total number of words:", word_count)
print("First 10 words:", first_10_words)
print("Words starting with a capital letter:", capital_words[:20])  # print first 20
```
