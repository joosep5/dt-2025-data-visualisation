```python
import urllib.request

# 1. Read text from URL
url = "https://raw.githubusercontent.com/joosep5/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/class_work/poem.txt"
text = urllib.request.urlopen(url).read().decode("utf-8")

# Show first 300 characters
print(text[:300])

# 2. Split text into words
words = text.split()

# 3. Print number of words
print("Total words in text:", len(words))

# 4. Print some selected words
print("First 10 words:", words[:10])
print("Last 10 words:", words[-10:])

# 5. Find words longer than 7 letters
long_words = [w for w in words if len(w) > 7]
print("Some long words:", long_words[:15])

# 6. Show every 10th word
print("Every 10th word:", words[::10][:20])
```

```python
from wordcloud import WordCloud

# Create a simple word cloud using the text from Task 1
wc = WordCloud(
    width=800,
    height=400,
    background_color="white",
    colormap="viridis"
).generate(text)

# Show the word cloud
wc.to_image()
```

```python
from PIL import Image
import numpy as np
import urllib.request

# Download a mask image
urllib.request.urlretrieve(
    "https://drive.google.com/uc?export=download&id=1o19a6yrRM6ikHr3FytSICT8nfdDA1qhK",
    "image1.png"
)

mask_image = np.array(Image.open("image1.png"))

wc_masked = WordCloud(
    mask=mask_image,
    background_color="green",
    colormap="plasma"
).generate(text)

wc_masked.to_image()
```

```python
# Together, the two word clouds show both:
# Which words are most important in the text (based on frequency so the bigger the word, the more it has been used).
# How these words can be presented visually to emphasize certain themes (through colors and masks).
```
