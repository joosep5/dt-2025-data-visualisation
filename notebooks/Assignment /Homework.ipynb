```python
import pandas as pd

# Laeme CSV andmed
df = pd.read_csv("Data.csv")

# Kontrollime esimesi ridu
df.head()
```

```python
import matplotlib.pyplot as plt

df["Salary"] = df["Salary"].str.replace(",", "").astype(float)

avg_salary = df.groupby("Office Branch")["Salary"].mean()

avg_salary.plot(kind="bar", figsize=(8,6), title="Average Salary by Office Branch")
plt.ylabel("Average Salary")
plt.show()
```

The column chart shows that the average salaries are highest in London and New York. Sydney and San Francisco have lower average salaries compared to these two branches. This suggests that location plays a role in compensation differences.

```python
office_counts = df["Office Branch"].value_counts()

office_counts.plot(kind="pie", autopct="%1.1f%%", figsize=(6,6), title="Employee Distribution by Office Branch")
plt.ylabel("")
plt.show()
```

The pie chart indicates that most employees are based in London and New York, while Sydney and San Francisco have fewer employees. This highlights the larger workforce concentration in the London and New York offices.

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Age"], df["Salary"], s=df["Salary"]/1000, alpha=0.6)

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs. Salary (Bubble chart)")
plt.show()
```

The scatter plot shows that salary is not strictly dependent on age. Some younger employees (in their 30s) earn very high salaries, while some older employees earn relatively less. This suggests that position and role have a stronger influence on salary than age.

```python
import seaborn as sns

plt.figure(figsize=(8,6))
sns.boxplot(x="Office Branch", y="Salary", data=df)

plt.title("Salary Distribution by Office Branch (Boxplot)")
plt.show()
```

The box plot reveals that salary distribution varies significantly across branches. London and New York show a wider salary range and higher outliers, while Sydney and San Francisco have comparatively lower salaries. This highlights disparities in compensation practices across different offices.
